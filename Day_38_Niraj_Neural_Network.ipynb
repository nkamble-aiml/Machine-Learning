````python
import numpy as np
class SimpleNeuron:
    def __init__(self, no_of_inputs, weights=None, bias=None):


        self.no_of_inputs = no_of_inputs

        if weights is None:
            self.weights = np.random.rand(no_of_inputs) * 0.01
        else:
            self.weights = weights
        if bias is None:
            self.bias = np.zeros(1)
        else:
            self.bias = bias

        print("Inputs: ", no_of_inputs)
        print("Weights are : ", self.weights)
        print("Bias are : ", self.bias)

    def __process(self, inputs):
        print("inputs", inputs)
        z = np.dot(inputs, self.weights) + self.bias
        return z

    def __sigmoid(self, z):
        return 1 / (1 + np.exp(-z)), z

    def decide(self,inputs, threshold = 0.5):
      z_score = self.__process(inputs)
      prob = self.__sigmoid(z_score)
      print(prob[0])
      print(threshold)
      if prob[0] < threshold:
        return "You can touch"
      else:
        return "Hot!!!, Do not touch"


oneNeu = SimpleNeuron(no_of_inputs=1)
decision = oneNeu.decide(-0.1)
print("Decision: ", decision)
````

````python
class SimpleNeuralNetwork:
    def __init__(self, input_features, output_class, learning_rate=0.1):
        self.input_features = input_features
        self.output_class = output_class
        self.learning_rate= learning_rate
        # Weight array is created with random value based on input features and output class to be produced
        self.W = np.random.randn(self.input_features, self.output_class) * 0.1
        self.b = np.zeros((1, self.output_class))

        print("Random Weights: ", self.W)
        print("Random Bias: ", self.b)

    def forwardPropogation(self, features):
       self.X = np.array(features)
       self.Z = np.dot(self.X, self.W) + self.b
       self.A = self.softmax(self.Z)

       for sample in range(len(self.A)):
        print(f"Sample Probability {sample+1}: {self.A[sample]}")



    def softmax(self,Z):
      self.A = np.exp(Z) / sum(np.exp(Z))
      return self.A

    def logLoss(self,Y_TRUE_INDEX):
      print("Y_TRUE_INDEX",Y_TRUE_INDEX)

      m = len(self.Z)

      losses = -np.log(self.A[range(m),Y_TRUE_INDEX])
      print("Losses: ", losses)

      for i in range(len(self.A)):
        print(f"For Sameple {i+1} :: -log( {self.A[i,Y_TRUE_INDEX[i]]} ) = {losses[i]}" )

      avg_loss = np.mean(losses)
      print("Average Loss ::", avg_loss)
      return losses

    def backwardPropogartion(self, Y_TRUE_INDEX):

      m = len(Y_TRUE_INDEX)
      print("m:",m)
      actual_label = np.zeros((m,self.output_class))
      actual_label[range(m),Y_TRUE_INDEX] = 1

      print("***" * 10)
      print("backward Propogation")
      print("Actual Value",actual_label)
      print("Probabilities Value",self.A)
      #weight error finding
      self.dZ = self.A - actual_label
      #find the Gradient Truth
      print("Delta Probabilities Value",self.dZ)
      self.dZ = self.dZ/m

      #calculate weight error
      self.dW = np.dot(self.X.T, self.dZ)
      print("Weight Error",self.dW)

      #calculate bias error
      self.db = np.sum(self.dZ, axis=0, keepdims=True)

      self.W = self.W - self.learning_rate * self.dW
      self.b = self.b - self.learning_rate * self.db
      print("Updated Weights",self.W)
      print("Updated Bias",self.b)

X = [[1.2,0.4],[-1.5,2.0]]
#0 - CAT
#1 - DOG
#2 - BIRD
Y_TRUE = [0,2]
simpleNN = SimpleNeuralNetwork(2,3,0.1)
simpleNN.forwardPropogation(X)
simpleNN.logLoss(Y_TRUE)
simpleNN.backwardPropogartion(Y_TRUE)
````
